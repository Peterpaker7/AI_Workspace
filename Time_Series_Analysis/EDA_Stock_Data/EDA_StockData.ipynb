```python
data = data[['Date', 'OpenPrice', 'HighPrice', 'LowPrice', 'ClosePrice']]
```

```python
data = data.rename(columns={
    'Open Price': 'Open',
    'High Price': 'High',
    'Low Price': 'Low',
    'Close Price': 'Close'
})
```

```python
data = data.set_index('Date')
```

```python

# Create a copy
ts = data.copy()

# Create dummy business-day dates
ts.index = pd.date_range(
    start="2024-01-01",
    periods=len(ts),
    freq="B"
)

# Seasonal decomposition
result = seasonal_decompose(
    ts["ClosePrice"],
    model="additive",
    period=30   # Adjust according to your data
)

# ---------------- Trend ----------------
plt.figure(figsize=(15,4))
plt.plot(result.trend, linewidth=2)
plt.title("Trend Component")
plt.xlabel("Dummy Date")
plt.ylabel("Trend")
plt.grid(True)
plt.show()

# ---------------- Seasonal ----------------
plt.figure(figsize=(15,4))
plt.plot(result.seasonal, linewidth=2)
plt.title("Seasonal Component")
plt.xlabel("Dummy Date")
plt.ylabel("Seasonality")
plt.grid(True)
plt.show()

# ---------------- Residual ----------------
plt.figure(figsize=(15,4))
plt.plot(result.resid, linewidth=2)
plt.title("Residual Component")
plt.xlabel("Dummy Date")
plt.ylabel("Residual")
plt.grid(True)
plt.show()
```

```python
from statsmodels.tsa.stattools import adfuller


def adf_test(series):

    result = adfuller(series.dropna())

    print("ADF Statistic :", result[0])
    print("p-value :", result[1])

    print("\nCritical Values:")
    for key, value in result[4].items():
        print(f"{key} : {value}")


    if result[1] <= 0.05:
        print("\nResult: Stationary")
    else:
        print("\nResult: Non-Stationary")
```

```python
adf_test(data["ClosePrice"])
```

```python

def stationarity_plot(series, window=30):

    rolling_mean = series.rolling(window=window).mean()
    rolling_std = series.rolling(window=window).std()


    plt.figure(figsize=(15,6))

    # Original series
    plt.plot(
        series,
        label="Original Data",
        linewidth=2
    )

    # Rolling mean
    plt.plot(
        rolling_mean,
        label=f"Rolling Mean ({window})",
        linewidth=2
    )

    # Rolling std
    plt.plot(
        rolling_std,
        label=f"Rolling Std ({window})",
        linewidth=2
    )


    plt.title(
        "Stationarity Check - Rolling Statistics"
    )

    plt.xlabel("Date")
    plt.ylabel("Value")

    plt.legend()

    plt.grid(True)

    plt.show()
```

```python
stationarity_plot(data["ClosePrice"])
```

```python
data["Difference"] = data["ClosePrice"].diff()

plt.figure(figsize=(15,5))

plt.plot(
    data["Difference"],
    label="Differenced Series"
)

plt.title(
    "First Order Differencing"
)

plt.legend()
plt.grid(True)

plt.show()
```

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf


fig, ax = plt.subplots(2,1,figsize=(12,8))


plot_acf(
    data["ClosePrice"].dropna(),
    lags=40,
    ax=ax[0]
)

ax[0].set_title("Autocorrelation Function (ACF)")


plot_pacf(
    data["ClosePrice"].dropna(),
    lags=40,
    ax=ax[1]
)

ax[1].set_title("Partial Autocorrelation Function (PACF)")


plt.tight_layout()
plt.show()
```
