```python
import pandas as pd
import datetime as dt
```

```python
stk_data=pd.read_csv('Tatacoffee13_21.csv')
```

```python
stk_data["Date"] = pd.to_datetime(stk_data["Date"])
```

```python
stk_data = stk_data[
    (stk_data["Date"] >= pd.Timestamp("2020-01-01")) &
    (stk_data["Date"] <= pd.Timestamp("2021-12-31"))
]
```

```python
#To pick a particular Column (Use [[]])
stk_data=stk_data[['Open','High','Low','Close']]
```

```python
column='Close'
```

```python
"""Time Series is a Univarinate , We are able to create a model for that one particular column (Date--> in the dataset is the TImeSeries data
    close-->Timeseries data,Open,High,Low)"""
```

```python
from sklearn.preprocessing import MinMaxScaler
Ms=MinMaxScaler()
data1=Ms.fit_transform(stk_data[[column]])
print('Len : ',data1.shape)
```

```python
# In TimeSeries Analysis we need a Standardsation or to Normalaisation we gonna use MinMaxScaler
```

```python
# training and Testset data
training_size=round(len(data1)*0.80) #80 Percentage 
print(training_size)
x_train=data1[:training_size]
x_test=data1[training_size:]
print('X_Train Length : ',x_train.shape)
print('X_Test Length : ',x_test.shape)
y_train=data1[:training_size]
y_test=data1[training_size:]
print('y_train Length : ', y_train.shape)
print('y_test length : ',y_test.shape)
```

```python
''' In time Series Analysis the Test&Training data should be split seriesly

In Machine Learning it will be random but in time series it should be series (Ordered)'''
```

```python
#Model Creation 
from sklearn.metrics import mean_squared_error # Metric 
trends=['n','t','c','ct'] # Parameters 
lags=[1,2,3,4]
from statsmodels.tsa.ar_model import AutoReg #Algo Used for Model Cretation
for i in lags:
    for td in trends:
        model=AutoReg(x_train,lags=i,trend=td)
        model_fit=model.fit()
        #Makes Predicts
        y_pred=model_fit.predict(len(x_train),len(data1)-1)
        from Auto_Regression_SF import sf
        print('Lag={},Trend={}'.format(i,td))
        sf.rmsemape(y_test,y_pred)
        print("***********")
        
```
