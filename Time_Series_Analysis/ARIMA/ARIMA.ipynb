```python
stk_data=pd.read_csv('Tatacoffee13_21.csv')
```

```python
stk_data=stk_data[['Open','High','Low','Close']]
stk_data
```

```python
column='Close'
```

```python
from sklearn.preprocessing import MinMaxScaler
Ms=MinMaxScaler()
data1=Ms.fit_transform(stk_data[[column]])
print('Len : ',data1.shape)
```

```python
training_size=round(len(data1)*0.80)
print(training_size)
x_train=data1[:training_size]
x_test=data1[training_size:]
print('x_train length : ',x_train.shape)
print('y_train length : ',x_test.shape)
y_train=data1[:training_size]
y_test=data1[training_size:]
print('y_train Length :',y_train.shape)
print('y_test Length : ',y_test.shape)
```

```python
order=[(1,1,1),(1,1,2),(2,3,1),(2,2,2)]
for i in order:
    from statsmodels.tsa.arima.model import ARIMA
    model=ARIMA(data1,order=i)
    model_fit=model.fit()
    y_pred=model_fit.predict(0,len(data1)-1)
    from ARIMA import sf
    sf.rmsemape(data1,y_pred)
```

```python
forecast=model_fit.predict(len(data1),len(data1)+3)
```

```python
forecast
```

```python
from ARIMA import sf
fTestNormTable=sf.conversionSingle(forecast,['closefore'])
forecast_stock_price_test_ori=Ms.inverse_transform(fTestNormTable)
forecast_stock_price_test_oriF=sf.conversionSingle(forecast_stock_price_test_ori,['closefore'])
```
