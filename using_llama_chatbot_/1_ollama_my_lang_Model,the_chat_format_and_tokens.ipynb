```python
import ollama
```

```python
def get_completion(prompt,model='llama3.2:1b'):
    response=ollama.chat(model=model,messages=[{'role':'user','content':prompt}])
    return response['message']['content']
```

```python
response=get_completion('What is the captial of tamil nadu?')
print(response)
```

```python
response=get_completion('Take the letters in Mukesh and reverse them and lollipop too')
print(response)
```

```python
response=get_completion('Take the letters in l-o-l-l-i-p-o-p and reverse them')
print(response)
```

```python
def get_completion_from_messages(messages,model='llama3.2:1b'):
    response=ollama.chat(model=model,messages=messages)
    return response['message']['content']
```

```python
messages = [
    {'role':'system','content':'You are an assistant who responds in the style of Dr Sesus.'},
    {'role':'user','content':'write me a very short poem about a happy carrot'}
]
response = get_completion_from_messages(messages)
print(response)
```

```python
#Length restriction example
messages=[
    {'role':'system','content':'All your response must be one sentence long.'},
    {'role':'user','content':'Write me a a story about a happy carrot'}
]
response=get_completion_from_messages(messages)
print(response)
```

```python
#Combined example 
messages=[
    {'role':'system','content':"""You are an assistant who response in the style of Dr.Suss
    All your response must be one sentence long."""},
    {'role':'user','content':"Write me a story about a happy carrot"}
]
response=get_completion_from_messages(messages)
print(response)
```

```python
def get_completion_and_token_count(meesgaes,model='llama3.2:1b'):
    response=ollama.chat(model,messages=messages)

    #ollama doesn't return exact token counts , so we estimate 
    formatted_prompt='\n'.join([msg['content'] for msg in messages])

    token_dict={
        'prompt_tokens':len(formatted_prompt.split()), #approximate count
        'completion_tokens':len(response['message']['content'].split()),
        'total_tokens':len(formatted_prompt.split())+len(response['message']['content'].split()),
    }

    return response['message']['content'],token_dict
```
