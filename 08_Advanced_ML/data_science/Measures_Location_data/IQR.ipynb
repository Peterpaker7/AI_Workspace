```python
Quan=[]
Qual=[]
```

```python
for columnName in dataset.columns:
    if dataset[columnName].dtype=="O":
        Qual.append(columnName)
    else:
        Quan.append(columnName)
```

```python
Quan
```

```python
descrptive=pd.DataFrame(index=["IQR",'lesser','1.5rule',"greater"],columns=Quan)
descrptive
```

```python
for columnName in Quan:
    descrptive[columnName]["IQR"]=dataset.describe()[columnName]['75%'] - dataset.describe()[columnName]['25%']
    descrptive[columnName]["1.5rule"]=1.5*descrptive[columnName]["IQR"]
    descrptive[columnName]["Lesser"]=dataset.describe()[columnName]['25%']-descrptive[columnName]["1.5rule"]
    descrptive[columnName]["Greater"]=dataset.describe()[columnName]['75%']+descrptive[columnName]["1.5rule"]
    
```

These columns are numeric features of candidates:

• ssc_p – Secondary school percentage
• hsc_p – Higher secondary percentage
• degree_p – Degree percentage
• etest_p – Employability test percentage
• mba_p – MBA percentage
• salary – Offered salary

Then you see rows like:

• IQR values
• 1.5×IQR lower bound (“lesser”)
• 1.5×IQR upper bound (“greater”)

That tells us this table is not raw data.
It’s a statistical summary used for outlier detection.

Now, what the IQR method means (plain language)
IQR = Q3 − Q1
It measures the middle 50% of the data.

Using IQR, we define:
• Lower bound = Q1 − 1.5 × IQR
• Upper bound = Q3 + 1.5 × IQR

Anything outside this range is considered an outlier.
