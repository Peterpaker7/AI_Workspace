```python
import pandas as pd
dataset=pd.read_csv("Placement_Data_Full_Class.csv")
```

```python
#creating the function named get_pdf_probability which will recived dataset and start and end range 
def get_pdf_probability(dataset,startrange,endrange):
    from matplotlib import pyplot as plt #for graph purpose (helps to plot)  
    from scipy.stats import norm # for stats thats curve (calculation)
    import seaborn as sns #Visuallisation that plot (cretaing it )

    ax=sns.distplot(dataset,kde=True,kde_kws={"color":"blue"},color="Green") #Creating that plot to see it 
    #kde--->Represent the Graph(line) ; kde_kws--> color for line ; color --> color for (bar)(histograph)
    pyplot=plt.axvline(startrange,color="Red") # axvline ---> denotes vertical line (that axVline that V denotes vertical ) (and this is for starting line)
    pyplot=plt.axvline(endrange,color="Red") #exactly same but its for end line 

    #Genrate Sample 
    Sample=dataset

    #Calculate Parameters 

    sample_mean=Sample.mean() #finding mean from the dataset
    sample_std=Sample.std() #finding standard deviation from the dataset

    print("Mean=%3f, StandardDevation=%3f"%(sample_mean,sample_std))

    #define the Distrubution

    dist=norm(sample_mean,sample_std) #normal disrubution caluculation in the mean and std dataset

    #Sample Probabilites for a range of outcomes 

    values=[value for value in range (startrange,endrange)]
    probabilites=[dist.pdf(values) for value in values]
    prob=sum(probabilites)
    print('The area between range({},{}):{}'.format(startrange,endrange,prob))

    return prob

```
