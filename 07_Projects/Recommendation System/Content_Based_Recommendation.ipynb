```python
# Function - Create a Vector list for whole user with to movie vector

def user_vector(ratings,movie_vector):
    vector=[]
    for user in range(1,944): # Each UserID to create vector
        u1=ratings[ratings['user_id']==user][['movie_id','ratings']]
        u1_list=[]
        for i in u1['movie_id']: #Each user watched movie genere vector 
            f=movie_vector[i].values
            u1_list.append(f)
        uu1=pd.DataFrame(u1_list,columns=range(1,19))
        vec=[]
        for j in range(1,19): #sum up the each genere of user 
            g=sum(uu1[j])
            vec.append(g/len(u1))
        vector.append(vec)
    return vector
```

```python
vector=user_vector(ratings,movie_vector)
```

```python
# Converting into Dataset

users_vector=pd.DataFrame(vector)
```

```python
#Dot Matrix for User_Vectors & Movie vector for the content vector

content_vector=np.dot(users_vector,movie_vector)
```

```python
#DataFrame for Content Vector

content_vector=pd.DataFrame(content_vector)
```

```python
#Transpose for the User Vector Access

content_vectors=content_vector.T
content_vectors
```

```python
# Function

def content_based(ratings,content_vectors,userid,thresh):
    u2=ratings[ratings['user_id']==userid]['movie_id'].values
    val=content_vectors[content_vectors[userid]>(content_vectors[userid].max()-thresh)][2]
    recom=[]
    for i in val.index:
        if(i in u2):
            pass
        else:
            recom.append(i)
    return recom
```
