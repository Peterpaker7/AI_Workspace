```python
import pandas as pd  

data = {'Name': ['Alice', 'Bob', 'Charlie'], 
        'Age': [25, 30, 35],  
        'Gender': ['F', 'M', 'M']}  

df = pd.DataFrame(data)

average_age = df['Age'].mean()

print(f"Average Age: {average_age}")
```

```python
import numpy as np
from scipy.stats import zscore

data = [1, 2, 3, 100, 5, 6, 7, 8, 9]
z_score=zscore(data)

outliers = [i for i,j in zip(data,z_score) if abs(j) > 2]
```

```python
import numpy as np
from scipy.stats import zscore

data = [1, 2, 3, 100, 5, 6, 7, 8, 9]
median = np.median(data)
z_score=zscore(data)

data = [median if abs(z_score[i]) > 2 else x for i, x in enumerate(data)]
```
