```python
def get_pdf_probability(dataset,startrange,endrange):
    from matplotlib import pyplot as plt
    from scipy.stats import norm
    import seaborn as sns 
    ax=sns.distplot(dataset,kde=True,kde_kws={'color':'blue'},color="Green")
    pylot=plt.axvline(startrange,color='Red')
    pyplot=plt.axvline(endrange,color='Red')
    #genrate sample
    sample=dataset
    #calculate parameters 
    sample_mean=sample.mean()
    sample_std=sample.std()
    print('Mean=%.3f , Standard Deviation=%.3f' % (sample_mean,sample_std))
    #define the distrubiton
    dist=norm(sample_mean,sample_std)

    #sample probablilties for a range of outcomes
    values=[value for value in range(startrange,endrange)]
    probablilties=[dist.pdf(value) for value in values]
    prob=sum(probablilties)
    print("The Area between range ({},{}):{}".format(startrange,endrange,prob))
    return prob
    
```

```python
get_pdf_probability(dataset["ssc_p"],50,70)
```

```python
def stdNBgraph(dataset):
    #convert to standard Normal Distribution 

    import seaborn as sns 
    mean=dataset.mean()
    std=dataset.std()

    values=[i for i in dataset]

    z_score=[((j-mean)/std)for j in values]

    sns.distplot(z_score,kde=True)
    sum(z_score)/len(z_score)
```

```python
stdNBgraph(dataset["ssc_p"])
```
