Scenario 2: Student Score Analysis

A teacher is analyzing the mathematics scores of students in her class. The scores are:
 [45, 50, 55, 60, 60, 62, 63, 65, 90, 95]
 
Question 1:
 Calculate the mean, median, and mode of the scores.
 
Question 2:
 Explain why the median might be a better representation than the mean in this case


```python
from scipy import stats
math_score=[45, 50, 55, 60, 60, 62, 63, 65, 90, 95]
math_array=np.array(math_score)
M_Mean=math_array.mean()
M_Median=np.median(math_array)
m_mode_result=stats.mode(math_array,keepdims=True)
M_Mode=int(m_mode_result.mode)
```

```python
print('Mean : ',M_Mean)
print('Median : ',M_Median)
print('M_Mode : ',M_Mode)
```

Scenario 3: Grocery Store Customer Analysis

A grocery store manager tracks how many customers visit the store daily for a month:
 [5, 10, 8, 15, 20, 5, 12, 14, 10, 18]
    
Question 1:
 Create a frequency distribution table for this data.


```python
import pandas as pd
customer_vist=[5, 10, 8, 15, 20, 5, 12, 14, 10, 18]
df=pd.DataFrame({'Customers':customer_vist})
#Count Frequency
freq_table=df['Customers'].value_counts().sort_index()
# Convert to DataFrame for better display
freq_df = pd.DataFrame({
    'Number of Customers': freq_table.index,
    'Frequency': freq_table.values
})

# Display the table
print("Frequency Distribution Table:")
print(freq_df.to_string(index=False))
```
